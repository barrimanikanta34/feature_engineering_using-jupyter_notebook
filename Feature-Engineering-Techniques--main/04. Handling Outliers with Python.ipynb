```python
# as we know that IQR is nothing but Q3 subtrcation with Q1
IQR = Q3-Q1
print(IQR)
```

```python
# FOR OUTER LIST WE NEED TO FIND HERE THE LOWER FENCE 
# because anything that is greater than higher fence will be considered as an outlier in this list 
# and anything that is lower then the lower fence than that is called as an outlier
lower_fence=Q1-1.5*(IQR)  #Formula for lower_fence 
higher_fence=Q3+1.5*(IQR)
```
