## Nominal/OHE Encoding 

One hot encoding, also known as nominal encoding, is a techniqueused to represent categorical data as numerical data, which is more suitable for machine learning algorithms. In this technique, each category is represented as a binary vector where each bit corresponds to a unique category. For example , if we have a categorical value 'Color' with three possible value (red,green,blue), we can represent the one hot encoding as follows :

1. Red:[1,0,0]
2. Green [0,1,0]
3. Blue[0,0,1]

```python
# process of converting Categorical features into Numerical Values , so that the model will be able to understand for a specific purpose .
```

```python
# in one hot encoding there is a major setback that if we have 100 types of categories then we need to create 100 no.of features.
# this is sparse matrix also. (Overfitting)
# so we can't be ablw to use one hot encoding when we have maximum variance.
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# create a simple dataframe 
df = pd.DataFrame({
    'color' : ['red','blue','green','green','red','blue']
}
)
```

```python
df.head()
```

```python
## create ab instance of one hot encoder 
encoder=OneHotEncoder()
```

```python
## perform fit and transform
encoded=encoder.fit_transform(df[['color']]).toarray() # here we converted and we cab see that it is basically based upon the alphabetical order 
## first should be Blue, then Green after that Red.
## this is how we probably create this.
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
## so in this way we are basically converting a categorical features into numerical features.
```

```python
## suppose you have got a new data then how it will happen ? 
encoder.transform([['blue']]).toarray()
# for new data
```

```python
pd.concat([df,encoder_df],axis=1)
```

## Label Encoding 

Label encoding and ordinal encoding are two techniques used to encode categorical data vs numerical data.

Label encoding involves assigning a unique numerical label to each category in the variable. The labels are usually assigned in alphabetical order on based on the frequency of the categories. For example, if we have a categorical variable 'color' with three possible values (red,green,blue), we can represent it using label encoding as follows:

1.Red : 1
2.Green: 2
3. Blue: 3

```python
# create a simple dataframe 
df = pd.DataFrame({
    'color' : ['red','blue','green','green','red','blue']
}
)
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['green']])
```

### Ordinal Encoding 

it is used to encode categorical data that have an intrinsic order or ranking. In this technique, eah category us assigned a numerical value based on its position in the order. For example, if we have a categorical variable "education level" with four possibles values (high school, college,graduate, post-graduate), We can represent it using ordinal coding as follows:

1.High School: 1
2.College:2
3.Graduate: 3
4.Post-Graduate : 4


```python
## ordinal encoding 
from sklearn.preprocessing import OrdinalEncoder
```

```python
# create a simple dataframe 
df = pd.DataFrame({
    'size': ['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
## create an instance of ordinal encoder and then fit_tranform 
encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

### Target Guided Ordinal Encoding 

It is a technique used to encode categorical variables based on their relationship with the target variable. This encoding technique is useful when we have a categorical variable with a large number of unique categories, and we want to use this variable as a feature in our machine learning model.

in Target Guided Ordinal Encoding , we replace each category in the categorical variable with a numerical value based on the mean or median of the target variable of this category. This creates a monotonic relationship between the categorical variable and the target variable, which can improve the predictive power of our model 

```python
import pandas as pd 

# create a sample dataframe with a categorical variable and a target c=variable 
df = pd.DataFrame({
    'city': ['New York','London','Paris','Tokyo','New York','Paris'],
    'price': [200,150,300,250,100,320]
})
```

```python
df # here price is a target variable based on this target varibale we will try to convert this categorical feature into 
# based on this target variable.
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
df[['city','city_encoded']]
```
