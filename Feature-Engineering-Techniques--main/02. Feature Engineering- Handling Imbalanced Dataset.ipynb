```python
# initially we will wrk with dataset where we will use 
# Classification - output- 2 categories - Binary classification 
# my output something like yes / no
# 900 yes and 100 yes with 1000 dataset 
# 9000:100 = 9;1 this is known as imbalanced datset 
# this type of imbalanced dataset will create a problem while we solving the problem of machine learning 
# the model that we creating which is responisble for predicting will get biased towards the maximum number of data
# so it is necessary to fix the imbalanced datset and we try to make syre thar both this data points becone equal.
# for this we will use 2 techniques 
# 1. UP sampling 2. Down Sampling 
# in up sampling we will increase the number of data points from 100 which is basically minority 
# in downsampling we will reduce the datapoints from here which is majority 
```

### Handling Imbalanced Dataset 
1.Up sampling 

2. Down Sampling 

```python
import numpy as np 
import pandas as pd 

# set the random seed for reproducibility 
np.random.seed(123)

# Create a Dataframe with 2 classes 
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
### Creating my Dataframe with imbalanced dataset
class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0] *n_class_0
})
class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1] *n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
print(df['target'].value_counts())
```

```python
## Up sampling 
df_minority=df[df['target'] == 1]
df_majority=df[df['target'] == 0]
```

```python
# here it basically creates and it extrapolates the point 
from sklearn.utils import resample 
df_minority_upsampled = resample(df_minority,replace=True,
        n_samples=len(df_majority),
        random_state=42) # Sample with replacement 
```

```python
df_minority_upsampled.shape ## this mean here i created more data points over here and all the dataset is basically giving 
# the output as 1.
```

```python
df_minority_upsampled.head() # confirmatory purpose 
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts() # from this we can see  that we increased the datapoints 
```

### for Down sampling 


```python
import numpy as np 
import pandas as pd 

# set the random seed for reproducibility 
np.random.seed(123)

# Create a Dataframe with 2 classes 
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

n_class_0,n_class_1

### Creating my Dataframe with imbalanced dataset
class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0] *n_class_0
})
class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1] *n_class_1
})

df = pd.concat([class_0,class_1]).reset_index(drop=True)

print(df['target'].value_counts())
```

```python
## Downsampling 
df_minority=df[df['target'] == 1]
df_majority=df[df['target'] == 0]
```

```python
# here it basically creates and it extrapolates the point 
from sklearn.utils import resample 
df_majority_downsampled = resample(df_majority,replace=False,
        n_samples=len(df_minority),
        random_state=42) # Sample with replacement 
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority,df_majority_downsampled])
```

```python
df_downsampled.target.value_counts()
# here downsampling is bad because here we are loosing our data points 
```
